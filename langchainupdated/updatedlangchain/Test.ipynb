```python
class Car:

    def __init__(self, brand, color):
        self.brand = brand
        self.color = color
        self.speed = 0

    def start(self):
        print(self.brand, "car has started")

    def accelerate(self, increase):
        self.speed = self.speed + increase
        print(self.brand, "Current speed:", self.speed)
```

```python
my_car = Car("Toyota", "Red")
my_car
```
